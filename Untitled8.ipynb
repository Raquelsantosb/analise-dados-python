```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

```python
# Criando um exemplo de DataFrame com dados ausentes
dados = {
    'Vendas': [1000, 1200, np.nan, 1500, 1300, np.nan, 1100],
    'Dia da Semana': ['Seg', 'Ter', 'Qua', 'Qui', 'Sex', 'Sáb', 'Dom'],
}

df = pd.DataFrame(dados)
```

```python
# Identificando os valores ausentes
valores_ausentes = df.isnull().sum()
print("Valores ausentes por coluna:")
print(valores_ausentes)
```

```python
# Visualização dos valores ausentes (Gráfico de dispersão)
plt.figure(figsize=(6,4))
plt.scatter(df.index, df['Vendas'], c=df['Vendas'].isnull(), cmap='coolwarm', label='Vendas')
plt.xlabel('Índice')
plt.ylabel('Vendas')
plt.title('Distribuição de Valores Ausentes')
plt.legend()
plt.show()
```

```python
# Preenchendo os valores ausentes com a média
df['Vendas'] = df['Vendas'].fillna(df['Vendas'].mean())
```

```python
# Visualizando o DataFrame após o preenchimento
print("\nDataFrame após preenchimento:")
print(df)
```
