```python
import random
```

```python
# Definindo as variáveis do experimento
variavel_independente = "Trilha de Estudos"  # Trilha de estudo escolhida
variavel_dependente = "Nível de Domínio"  # Nível de domínio adquirido
```

```python
# Definindo as trilhas de estudo disponíveis
trilhas_estudo = {
    "Leitura de Livros": [10, 20, 30],  # Horas sugeridas
    "Cursos Online": [15, 25, 40],
    "Projetos Prᴩcos": [20, 30, 50],
    "Tutoria/Apoio de Mentores": [25, 35, 45]
}
```

```python
# Função para simular o nível de domínio com base na trilha de estudo e tempo dedicado
def simula_dominio(trilha, horas_dedicadas):
    # Simulação simples: domínio aumenta com horas dedicadas e variação aleatória
    dominio = horas_dedicadas * random.uniform(0.7, 1.3)  # Fator aleatória para simular variação
    return round(dominio)
```

```python
# Executa o experimento
resultados = {}
for trilha, horas_sugeridas in trilhas_estudo.items():
    horas = random.choice(horas_sugeridas)  # Escolhe uma quantidade de horas sugerida aleatoriamente
    dominio = simula_dominio(trilha, horas)
    resultados[trilha] = (horas, dominio)
```

```python
# Exibe os resultados
print("Resultados do Experimento - Trilha de Estudos:")
print("---------------------------------------------")
print(f"{variavel_independente} | Horas Dedicadas | {variavel_dependente}")
print("---------------------------------------------")
for trilha, (horas, dominio) in resultados.items():
    print(f"{trilha} | {horas} horas | {dominio} pontos de domínio")
print("---------------------------------------------")
```

```python
# Analise os resultados
trilha_otima = max(resultados, key=lambda x: resultados[x][1])
print("\nAnálise dos Resultados:")
print(f"A trilha de estudo com o maior nível de domínio é{trilha_otima}'")
print(f"com {resultados[trilha_otima][1]} pontos de domínio.")
```
