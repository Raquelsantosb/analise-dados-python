```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
from google.colab import files
```

```python
# Função para ler o arquivo de texto e extrair os dados
def ler_arquivo_produtos(uploaded):
    # Pegar o nome do arquivo do dicionário 'uploaded'
    file_name = list(uploaded.keys())[0]
    df = pd.read_csv(file_name, sep='\t')
    return df
```

```python
# Function to display a colorful horizontal bar chart using seaborn
def exibir_grafico_seaborn(df):
    # Define the color palette
    sns.set(style="whitegrid")
    # Create the horizontal bar chart with hue as 'Produto'
    plt.figure(figsize=(8, 6))
    grafico = sns.barplot(x='Nível de Aceitação',
                          y='Produto',
                          data=df,
                          hue='Produto',
                          palette="coolwarm",
                          dodge=False, legend=False)
    # Add title and labels
    plt.title('Product Acceptance Level', fontsize=14)
    plt.xlabel('Acceptance Level')
    plt.ylabel('Product')
    # Display the chart
    plt.show()
```
