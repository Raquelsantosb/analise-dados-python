```python
import pandas as pd

from statsmodels.tsa.seasonal import seasonal_decompose

import matplotlib.pyplot as plt

import seaborn as sns
```

```python
df = pd.read_excel('/content/pim_569b562b05xlsx.xlsx')
df.head()
df.info()
```

```python
plt.plot(df['Month'], df['Value'])
plt.show()

# GRÁFICO DA PRODUÇÃO INDUSTRIAL MENSAL PIM-PF
```

```python
# Transformar a data em índice (apenas se 'Month' ainda for uma coluna)
if 'Month' in df.columns:
    df['Month'] = pd.to_datetime(df['Month'])
    df.set_index('Month', inplace=True)
```

```python
#salvar a decomposição no modelo aditivo

result = seasonal_decompose(df, model='additive')
result.plot()
plt.show()
```

```python
# decompõe a série de acordo com o modelo multiplicativo

result_mult = seasonal_decompose(df, model='multiplicative')
result_mult.plot()
plt.show()
```
