```python
import pandas as pd
```

```python
import numpy as np
```

```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
# simulando dados de uma pesquisa
np.random.seed(42) # para reprodutibilidade
dados = {
    'pergunta_1': np.random.choice(['sim', 'não', 'talvez'], size=100),
    'pergunta_2': np.random.randint(1, 5, size=100),
    'grupo': np.random.choice(['grupo A', 'grupo B'], size=100)
}

df = pd.DataFrame(dados)

# filtrando apenas o grupo A
df_grupo_a = df[df['grupo'] == 'grupo A']

# calculando a proporcao das respostas para o grupo A
df_perc_a = df_grupo_a['pergunta_1'].value_counts(normalize=True).mul(100).reset_index()
df_perc_a.columns = ['pergunta_1', 'percentual']

# visualizacao com searborn - percentual de pergunta 1 para o grupo A
plt.figure(figsize=(8, 6))
sns.set_style("whitegrid")

# Agora usamos a variavel `hue` para controlar as cores das barras
sns.barplot(x='pergunta_1', y='percentual', hue='pergunta1', data=df_perc_a,
palette='viridis', dodge=False, legend=False)

# ajustes de titulos e eixos
plt.title('percentual de respostas pergunta 1 - grupo A', fontsize=16)
plt.xlabel('resposta pergunta 1', fontsize=12)
plt.ylabel('percentual (%)', fontsize=12)
plt.tight_layout()  # ajuste para evitar cortes
plt.show()

```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
# Simulando dados de uma pesquisa
np.random.seed(42)  # Para reprodutibilidade
dados = {
    'pergunta_1': np.random.choice(['sim', 'não', 'talvez'], size=100),
    'Pergunta_2': np.random.randint(1, 5, size=100),
    'Grupo': np.random.choice(['Grupo A', 'Grupo B'], size=100)
}

df = pd.DataFrame(dados)

# Filtrando apenas o Grupo B
df_grupo_b = df[df['Grupo'] == 'Grupo B']

# Calculando a proporção das respostas para o Grupo B

df_perc_b = df_grupo_b['pergunta_1'].value_counts(normalize=True).mul(100).reset_index()
df_perc_b.columns = ['pergunta_1', 'Percentual']

# Visualização com Seaborn - Percentual de Pergunta 1 para o Grupo B
plt.figure(figsize=(8, 6))
sns.set_style("whitegrid")

# Usando a variável `hue` para controlar as cores das barras
sns.barplot(x='pergunta_1', y='Percentual', hue='pergunta_1', data=df_perc_b, palette='magma', dodge=False, legend=False)

# Ajustes de títulos e eixos
plt.title('Percentual de Respostas Pergunta 1 - Grupo B', fontsize=16)
plt.xlabel('Resposta Pergunta 1', fontsize=12)
plt.ylabel('Percentual (%)', fontsize=12)
plt.tight_layout()  # Ajuste para evitar cortes
plt.show()
```
